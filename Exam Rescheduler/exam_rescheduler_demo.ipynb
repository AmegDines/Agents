# 🎓 Exam Rescheduler AI Agent Demonstration

This notebook demonstrates an autonomous AI agent capable of rescheduling university exams while adhering to strict constraints:

1.  **No Student Clashes**: No student should have two exams in the same slot.
2.  **Room Capacity**: Rooms must fit all students with a one-seat gap (Approx $2n-1$ seats).
3.  **Faculty Availability**: All invigilators must be available.
4.  **Safety First**: No irreversible actions without human approval.
5.  **Efficiency**: Total tool calls must stay under 40.

### System Architecture
- **Brain (LLM)**: Decides the next action based on current state and context.
- **Hands (Tools)**: Interfaces with a mock Campus ERP.
- **Memory (Context)**: Stores gathered data and the proposed plan.
- **Audit (Trace)**: Logs every single decision and result.
